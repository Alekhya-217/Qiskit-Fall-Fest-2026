```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
from IPython.display import display

simulator = AerSimulator()
print("Setup completed!")
```

```python
qc_a = QuantumCircuit(1, 1)

qc_a.x(0)
qc_a.measure(0, 0)

display(qc_a.draw('mpl'))

result_a = simulator.run(qc_a, shots=100).result()
counts_a = result_a.get_counts()

print("Circuit A results:", counts_a)
display(plot_histogram(counts_a))
```

```python
qc_b = QuantumCircuit(1, 1)

qc_b.x(0)
qc_b.x(0)
qc_b.measure(0, 0)

display(qc_b.draw('mpl'))

result_b = simulator.run(qc_b, shots=100).result()
counts_b = result_b.get_counts()

print("Circuit B results:", counts_b)
display(plot_histogram(counts_b))
```

```python
qc_c = QuantumCircuit(1, 1)

qc_c.h(0)
qc_c.measure(0, 0)

display(qc_c.draw('mpl'))

result_c = simulator.run(qc_c, shots=100).result()
counts_c = result_c.get_counts()

print("Circuit C results:", counts_c)
display(plot_histogram(counts_c))
```

```python
display(plot_histogram(
    [counts_a, counts_b, counts_c],
    legend=[
        'Circuit A - One X gate',
        'Circuit B - Two X gates',
        'Circuit C - One H gate'
    ]
))
```

- Circuit B: Two X gates return the qubit to state 0.

### Observation
Circuit A produces 1 in all 100 shots. Circuit B produces 0 in all 100 shots. Circuit C produces approximately equal numbers of 0 and 1.

### Comparison and Conclusion
