```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
from IPython.display import display

simulator = AerSimulator()
```

```python
qc = QuantumCircuit(1, 1)

qc.ry(1.0472, 0)
qc.measure(0, 0)

display(qc.draw('mpl'))
```

```python
result = simulator.run(qc, shots=100).result()
counts = result.get_counts()

print("Measurement results:", counts)
display(plot_histogram(counts))
```

## Challenge 2: Make the Quantum Coin Biased

### Prediction
The outcome 0 is expected to occur approximately 75% of the time, while 1 is expected to occur approximately 25% of the time.

### Observation
The histogram shows that 0 occurs more frequently than 1. The exact counts may vary over 100 shots.

### Conclusion
The RY gate changes the quantum state and creates unequal measurement probabilities. Therefore, the circuit behaves like a biased quantum coin.
