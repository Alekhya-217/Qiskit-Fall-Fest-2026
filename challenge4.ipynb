```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
from IPython.display import display

simulator = AerSimulator()

print("Libraries imported successfully!")
```

```python
qc_a = QuantumCircuit(2, 2)

qc_a.h(0)
qc_a.cx(0, 1)
qc_a.measure([0, 1], [0, 1])

display(qc_a.draw('mpl'))
```

```python
result_a = simulator.run(qc_a, shots=100).result()
counts_a = result_a.get_counts()

print("Part A results:", counts_a)
display(plot_histogram(counts_a))
```

```python
qc_b = QuantumCircuit(2, 2)

qc_b.h(0)
qc_b.cx(0, 1)
qc_b.x(0)
qc_b.measure([0, 1], [0, 1])

display(qc_b.draw('mpl'))

result_b = simulator.run(qc_b, shots=100).result()
counts_b = result_b.get_counts()

print("Part B results:", counts_b)
display(plot_histogram(counts_b))
```

```python
display(plot_histogram(
    [counts_a, counts_b],
    legend=['Part A - Correlated', 'Part B - Opposite']
))
```

## Challenge 4: Quantum Correlation Switch

### Predictions
- Part A: Outcomes 00 and 11 are expected to dominate.
- Part B: Outcomes 01 and 10 are expected to dominate.

### Comparison and Observation
Part A produces correlated outcomes, mainly 00 and 11. Part B produces opposite outcomes, mainly 01 and 10, because an X gate flips one qubit.

### Conclusion
Adding an X gate changes the relationship between the two qubits' measurement outcomes. The histograms demonstrate the difference between correlated and opposite outcomes.
